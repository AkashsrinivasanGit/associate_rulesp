```python
Uk_data=Online_retail_data[Online_retail_data["Country"] =="United Kingdom"]
Uk_data
```

```python
Uk_data.isna().sum()
```

```python
del Uk_data["CustomerID"]
```

```python
Uk_data.dropna(inplace=True)
```

```python
Uk_data.isna().sum()
```

```python
Uk_pivotte =pd.pivot_table(data =Uk_data,values="Quantity",index="InvoiceDate",columns="Description",aggfunc="sum").fillna(0)
Uk_pivotte
```

```python
Uk_pivotte_onehot=Uk_pivotte.applymap(lambda x:0 if x == 0 else 1)
Uk_pivotte_onehot
```

```python
Uk_data
```

```python
##data mining
```

```python
from mlxtend.frequent_patterns import apriori,association_rules
```

```python
frequent_associate=apriori(  df =Uk_pivotte_onehot ,min_support=0.02,use_colnames=True)
frequent_associate
```

```python
best_asso=association_rules(df=frequent_associate,metric='confidence', min_threshold=0.02)
best_asso
```

```python
best_asso.to_csv("best_antedecent_consequent.csv")
```
